### Ejercicio 2.5 · Unir las cestas de la compra

Un supermercado online guarda los productos de cada cesta en una lista independiente:

```python
cestas = [["pan", "leche"], ["huevos"], [], ["leche", "café", "pan"]]
```

7. Con una list comprehension **con dos ****`for`**, crea una única lista con todos los productos de todas las cestas.
8. Con una dictionary comprehension, crea un diccionario que indique cuántas veces aparece cada producto. Pista: puedes usar el método `count()` de las listas.
**Salida esperada**

```plain text
['pan', 'leche', 'huevos', 'leche', 'café', 'pan']

```

---


````python
# ==========================================
# Ejercicio 2.1 · Precios con IVA
# ==========================================
precios = [8.50, 12.00, 3.20, 45.90, 19.99, 7.00]

# Filtramos al final (if p > 10) y transformamos al principio aplicando el 21% y redondeando
precios_iva = [round(p * 1.21, 2) for p in precios if p > 10]
print(precios_iva)  # [14.52, 55.54, 24.19]


# ==========================================
# Ejercicio 2.2 · Clasificar paquetes
# ==========================================
pesos = [0.5, 3.2, 12.0, 1.1, 25.4, 7.8]

# Usamos if/else al principio porque queremos clasificar TODOS los paquetes, no filtrar
clasificacion = ["LIGERO" if p < 5 else "PESADO" for p in pesos]
print(clasificacion)
print(f"Paquetes pesados: {clasificacion.count('PESADO')}")


# ==========================================
# Ejercicio 2.3 · Ciudades sin repetir
# ==========================================
ciudades = ["Madrid", " madrid", "SEVILLA", "Valencia ", "sevilla", "Bilbao", "MADRID"]

# Set comprehension con llaves {}: strip() quita espacios y capitalize() formatea el texto
ciudades_unicas = {c.strip().capitalize() for c in ciudades}

print(sorted(ciudades_unicas))
print(f"Ciudades distintas: {len(ciudades_unicas)}")


# ==========================================
# Ejercicio 2.4 · Butacas libres en un cine
# ==========================================
butacas_libres = {"Sala 1": 0, "Sala 2": 34, "Sala 3": 6, "Sala 4": 0, "Sala 5": 120}

# 4. Lista con los nombres de las salas completas (0 butacas libres)
salas_completas = [sala for sala, libres in butacas_libres.items() if libres == 0]

# 5. Diccionario con el estado de cada sala (usando operador ternario encadenado)
estado_salas = {
    sala: ("COMPLETA" if libres == 0 else "ÚLTIMAS PLAZAS" if libres < 10 else "DISPONIBLE")
    for sala, libres in butacas_libres.items()
}

# 6. Diccionario de porcentaje de ocupación en salas NO completas (aforo total = 150)
ocupacion = {
    sala: round((150 - libres) / 150 * 100, 1)
    for sala, libres in butacas_libres.items()
    if libres > 0
}

print(salas_completas)
print(estado_salas)
print(ocupacion)


# ==========================================
# Ejercicio 2.5 · Unir las cestas de la compra
# ==========================================
cestas = [["pan", "leche"], ["huevos"], [], ["leche", "café", "pan"]]

# 7. Doble for: primero recorre cada cesta y luego cada producto dentro de esa cesta
todos_productos = [producto for cesta in cestas for producto in cesta]

# 8. Contamos apariciones. En Python 3.7+ los diccionarios mantienen el orden de inserción
conteo_productos = {prod: todos_productos.count(prod) for prod in todos_productos}

print(todos_productos)
print(conteo_productos)
````

### Ejercicio 3.6 · Funciones que devuelven funciones

La empresa cobra un precio por kilómetro distinto según la franja horaria. Escribe una función `crear_tarifa(precio_km)` que **devuelva otra función**. Esa función recibirá unos kilómetros y devolverá el precio del trayecto, redondeado a 2 decimales.

```python
tarifa_diurna = crear_tarifa(1.10)
tarifa_nocturna = crear_tarifa(1.35)

print([tarifa_nocturna(c["km"]) for c in carreras])
```

---


````python
from functools import reduce

carreras = [
    {"id": 1, "pasajero": "Raúl",    "km": 12.0, "importe": 18.5, "tarjeta": True},
    {"id": 2, "pasajero": "Sofía",   "km": 3.5,  "importe": 7.0,  "tarjeta": False},
    {"id": 3, "pasajero": "Iván",    "km": 28.0, "importe": 42.0, "tarjeta": True},
    {"id": 4, "pasajero": "Claudia", "km": 8.0,  "importe": 13.5, "tarjeta": True},
    {"id": 5, "pasajero": "Hugo",    "km": 15.0, "importe": 24.0, "tarjeta": False},
]

# ==========================================
# Ejercicio 3.1 · Funciones lambda
# ==========================================
# 9. Sumar 2.50 € de suplemento
con_suplemento = lambda importe: importe + 2.50

# 10. Comprobar si son 10 km o más
es_larga = lambda km: km >= 10

# 11. Ordenar de mayor a menor km (reverse=True) y extraer el nombre
carreras_ordenadas = sorted(carreras, key=lambda c: c["km"], reverse=True)
nombres_ordenados = [c["pasajero"] for c in carreras_ordenadas]

print(con_suplemento(18.5))
print(es_larga(3.5))
print(nombres_ordenados)


# ==========================================
# Ejercicio 3.2 · map()
# ==========================================
# 12. Lista de kilómetros
lista_km = list(map(lambda c: c["km"], carreras))

# 13. Textos formateados
textos_carreras = list(map(lambda c: f"Carrera {c['id']}: {c['importe']} €", carreras))

print(lista_km)
print(textos_carreras)


# ==========================================
# Ejercicio 3.3 · filter()
# ==========================================
# 14. Pagadas en efectivo (not c["tarjeta"]) -> mostramos pasajero
efectivo = [c["pasajero"] for c in filter(lambda c: not c["tarjeta"], carreras)]

# 15. Con tarjeta y >= 10 km -> mostramos id
tarjeta_largas = [c["id"] for c in filter(lambda c: c["tarjeta"] and c["km"] >= 10, carreras)]

print(efectivo)
print(tarjeta_largas)


# ==========================================
# Ejercicio 3.4 · reduce()
# ==========================================
# 16. Recaudación total (usamos 0 como valor inicial del acumulador)
recaudacion_total = reduce(lambda acc, c: acc + c["importe"], carreras, 0)

# 17. Carrera más larga (compara dos diccionarios de carrera y se queda con el de mayor km)
carrera_mas_larga = reduce(lambda c1, c2: c1 if c1["km"] > c2["km"] else c2, carreras)

# 18. Km en carreras pagadas con tarjeta combinando filter() y reduce()
carreras_tarjeta = filter(lambda c: c["tarjeta"], carreras)
km_tarjeta = reduce(lambda acc, c: acc + c["km"], carreras_tarjeta, 0)

print(f"Recaudación total: {recaudacion_total} €")
print(f"Carrera más larga: {carrera_mas_larga['pasajero']}")
print(f"Km pagados con tarjeta: {km_tarjeta}")


# ==========================================
# Ejercicio 3.5 · Funciones que reciben funciones
# ==========================================
def transformar_importes(carreras, operacion):
    # Aplica la función 'operacion' recibida por parámetro al importe de cada carrera
    return [operacion(c["importe"]) for c in carreras]

def a_dolares(x):
    return round(x * 1.08, 2)

print(transformar_importes(carreras, a_dolares))
print(transformar_importes(carreras, lambda x: round(x * 1.10, 2)))


# ==========================================
# Ejercicio 3.6 · Funciones que devuelven funciones
# ==========================================
def crear_tarifa(precio_km):
    # Clausura: la función interna recuerda 'precio_km'
    def calcular(km):
        return round(km * precio_km, 2)
    return calcular

tarifa_diurna = crear_tarifa(1.10)
tarifa_nocturna = crear_tarifa(1.35)

print(tarifa_diurna(10))
print(tarifa_nocturna(10))
print([tarifa_nocturna(c["km"]) for c in carreras])
````

### Ejercicio 4.5 · Iterativo frente a recursivo

Una persona ahorra dinero en una cuenta que da un **2 % de interés al año**. Escribe dos funciones que calculen el dinero que tendrá tras `n` años, partiendo de un capital inicial:

- `ahorro_iterativo(capital, anios)`, con un bucle;
- `ahorro_recursivo(capital, anios)`, sin bucles.
Redondea el resultado a 2 decimales al mostrarlo.

```python
print(round(ahorro_iterativo(1000, 5), 2))
print(round(ahorro_recursivo(1000, 5), 2))
```

**Salida esperada**

```plain text
1104.08
1104.08
```

**Pregunta:** ¿qué ocurriría con la versión recursiva si se llamara con `anios = 5000`?

---


````python
# ==========================================
# Ejercicio 4.1 · Sumar los dígitos de un código
# ==========================================
def suma_digitos(n):
    # Caso base: si el número tiene una sola cifra, la suma es él mismo
    if n < 10:
        return n
    # Caso recursivo: última cifra (n % 10) + suma del resto de cifras (n // 10)
    return (n % 10) + suma_digitos(n // 10)

print(suma_digitos(4528))  # 19
print(suma_digitos(7))     # 7


# ==========================================
# Ejercicio 4.2 · Total de una lista de gastos
# ==========================================
def total_gastos(gastos):
    # Caso base: lista vacía suma 0
    if not gastos:
        return 0
    # Caso recursivo: primer elemento + total del resto de la lista (slicing)
    return gastos[0] + total_gastos(gastos[1:])

print(total_gastos([12.5, 30, 7.5, 50]))  # 100.0
print(total_gastos([]))                   # 0


# ==========================================
# Ejercicio 4.3 · ¿Es un palíndromo?
# ==========================================
def es_palindromo(texto):
    # Caso base: un texto de 0 o 1 carácter siempre es palíndromo
    if len(texto) <= 1:
        return True
    # Si los extremos no coinciden, ya sabemos que NO es palíndromo
    if texto[0] != texto[-1]:
        return False
    # Caso recursivo: comprobamos la subcadena interior quitando el primero y el último
    return es_palindromo(texto[1:-1])

print(es_palindromo("reconocer"))  # True
print(es_palindromo("radar"))      # True
print(es_palindromo("python"))     # False


# ==========================================
# Ejercicio 4.4 · Cajas dentro de cajas
# ==========================================
envio = [2, [3, 1], [[4, 5], 2], [], 6]

# 19. Peso total
def peso_total(caja):
    total = 0
    for elemento in caja:
        if isinstance(elemento, list):
            total += peso_total(elemento)  # Si es una caja interior, llamada recursiva
        else:
            total += elemento              # Si es un artículo, sumamos su peso
    return total

# 20. Contar artículos
def contar_articulos(caja):
    cantidad = 0
    for elemento in caja:
        if isinstance(elemento, list):
            cantidad += contar_articulos(elemento)  # Si es caja, contamos lo que tenga dentro
        else:
            cantidad += 1                           # Si es número, cuenta como 1 artículo
    return cantidad

print(f"Peso total: {peso_total(envio)} kg")
print(f"Número de artículos: {contar_articulos(envio)}")


# ==========================================
# Ejercicio 4.5 · Iterativo frente a recursivo
# ==========================================
def ahorro_iterativo(capital, anios):
    for _ in range(anios):
        capital *= 1.02
    return capital

def ahorro_recursivo(capital, anios):
    # Caso base: 0 años transcurridos devuelve el capital sin cambios
    if anios == 0:
        return capital
    # Caso recursivo: aplicamos un año de interés y restamos 1 a los años pendientes
    return ahorro_recursivo(capital * 1.02, anios - 1)

print(round(ahorro_iterativo(1000, 5), 2))  # 1104.08
print(round(ahorro_recursivo(1000, 5), 2))  # 1104.08
````

````python
# ==========================================
# Ejercicio 5.1 · Artículos de una tienda de deportes
# ==========================================
class Articulo:
    def __init__(self, nombre, precio, stock=0):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock

    def valor_inventario(self):
        return self.precio * self.stock

    def vender(self, cantidad):
        if self.stock >= cantidad:
            self.stock -= cantidad
            return True
        return False

    def reponer(self, cantidad):
        self.stock += cantidad

    def __str__(self):
        return f"{self.nombre} | {self.precio} € | {self.stock} uds."

a1 = Articulo("Balón de fútbol", 25.0, 10)
a2 = Articulo("Cinta de pádel", 9.5)

print(a1)
print(a2)
print(a1.vender(3))
print(a2.vender(1))
a2.reponer(20)
print(a1)
print(a2)
print(a1.valor_inventario(), a2.valor_inventario())


# ==========================================
# Ejercicio 5.2 · Cuentas de un banco
# ==========================================
class Cuenta:
    total_cuentas = 0
    COMISION_RETIRADA = 1.5

    def __init__(self, titular, saldo=0.0):
        self.titular = titular
        self._saldo = float(saldo)
        Cuenta.total_cuentas += 1

    def ingresar(self, importe):
        if importe > 0:
            self._saldo += importe

    def retirar(self, importe):
        coste_total = importe + Cuenta.COMISION_RETIRADA
        if self._saldo >= coste_total:
            self._saldo -= coste_total
            return True
        return False

    def consultar_saldo(self):
        return self._saldo

    @classmethod
    def cuentas_creadas(cls):
        return cls.total_cuentas

    def __repr__(self):
        return f"Cuenta('{self.titular}', {self._saldo})"

c1 = Cuenta("Laura", 100.0)
c2 = Cuenta("Diego")

c1.ingresar(50)
c1.ingresar(-20)
print(c1.retirar(48.5))
print(c2.retirar(10))
print(c1, c2)
print("Cuentas creadas:", Cuenta.cuentas_creadas())


# ==========================================
# Ejercicio 5.3 · Carrito de la compra (composición)
# ==========================================
class Carrito:
    def __init__(self):
        self.lineas = []  # Lista de tuplas (articulo, cantidad)

    def agregar(self, articulo, cantidad):
        self.lineas.append((articulo, cantidad))

    def total(self):
        return sum(art.precio * cant for art, cant in self.lineas)

    def resumen(self):
        for art, cant in self.lineas:
            subtotal = art.precio * cant
            print(f"{cant} x {art.nombre} = {subtotal} €")
        print(f"TOTAL: {self.total()} €")

carrito = Carrito()
carrito.agregar(Articulo("Balón de fútbol", 25.0, 10), 3)
carrito.agregar(Articulo("Cinta de pádel", 9.5, 20), 2)
carrito.agregar(Articulo("Botella deportiva", 15.0, 5), 1)
carrito.resumen()


# ==========================================
# Ejercicio 5.4 · Lista de objetos
# ==========================================
catalogo = [
    Articulo("Esterilla", 30.0, 12),
    Articulo("Comba", 12.0, 0),
    Articulo("Bicicleta estática", 180.0, 3),
    Articulo("Mancuernas", 45.0, 0),
]

# 21. Nombres de los artículos sin stock
print([a.nombre for a in catalogo if a.stock == 0])

# 22. Diccionario nombre -> valor_inventario()
print({a.nombre: a.valor_inventario() for a in catalogo})

# 23. Catálogo ordenado por precio de menor a mayor
for a in sorted(catalogo, key=lambda x: x.precio): print(a)
````

````python
# ==========================================
# Ejercicio 6.1 · Sumar y comparar importes
# ==========================================
class Importe:
    def __init__(self, cantidad):
        self.cantidad = float(cantidad)

    # Operador +
    def __add__(self, otro):
        return Importe(round(self.cantidad + otro.cantidad, 2))

    # Operador ==
    def __eq__(self, otro):
        return round(self.cantidad, 2) == round(otro.cantidad, 2)

    # Operador <
    def __lt__(self, otro):
        return self.cantidad < otro.cantidad

    # Conversión a str()
    def __str__(self):
        return f"{self.cantidad:.2f} €"

    # Conversión a float()
    def __float__(self):
        return self.cantidad

a = Importe(19.99)
b = Importe(5.01)
c = Importe(25)

total = a + b
print(total)
print(total == c)
print(a < b)
print(float(a) * 2)


# ==========================================
# Ejercicio 6.2 · Lista de reproducción
# ==========================================
class ListaReproduccion:
    def __init__(self, nombre):
        self.nombre = nombre
        self.canciones = []

    # len(lista)
    def __len__(self):
        return len(self.canciones)

    # "titulo" in lista
    def __contains__(self, cancion):
        return cancion in self.canciones

    # lista += "titulo"
    def __iadd__(self, cancion):
        self.canciones.append(cancion)
        return self  # Imprescindible devolver self en __iadd__

    # str(lista)
    def __str__(self):
        return f"{self.nombre} ({len(self)} canciones)"

viaje = ListaReproduccion("Viaje")
viaje += "Canción A"
viaje += "Canción B"
viaje += "Canción C"

print(viaje)
print(len(viaje))
print("Canción B" in viaje)
print("Canción Z" in viaje)


# ==========================================
# Ejercicio 6.3 · Precios escritos a mano
# ==========================================
class Precio:
    def __init__(self, valor):
        self.valor = float(valor)

    @staticmethod
    def desde_texto(texto):
        # Quitamos el símbolo €, los espacios y cambiamos la coma por punto decimal
        limpio = texto.replace("€", "").strip().replace(",", ".")
        return Precio(float(limpio))

    def __float__(self):
        return self.valor

    def __bool__(self):
        return self.valor != 0.0

    def __str__(self):
        return f"{self.valor:.2f} €"

carta = ["12,50 €", "0,00 €", "7,25 €", "18,00 €"]
precios = [Precio.desde_texto(t) for t in carta]

print([float(p) for p in precios])
print("Total carta:", sum(float(p) for p in precios))
print("Platos gratuitos:", len([p for p in precios if not p]))
print(precios[2])
````

````python
class Suscripcion:
    def __init__(self, usuario, cuota_base):
        self.usuario = usuario
        self.cuota_base = cuota_base

    def precio_mensual(self):
        return float(self.cuota_base)

    def ficha(self):
        tipo = type(self).__name__
        return f"{self.usuario} ({tipo}): {self.precio_mensual():.2f} €"

class Familiar(Suscripcion):
    def __init__(self, usuario, cuota_base, perfiles_extra):
        super().__init__(usuario, cuota_base)
        self.perfiles_extra = perfiles_extra

    def precio_mensual(self):
        return super().precio_mensual() + (3 * self.perfiles_extra)

class Anual(Suscripcion):
    def precio_mensual(self):
        return super().precio_mensual() * 0.85

class ConAnuncios(Suscripcion):
    def precio_mensual(self):
        return super().precio_mensual() * 0.60

suscripciones = [
    Suscripcion("Carlos", 12),
    Familiar("Lucía", 12, 2),
    Anual("Sergio", 12),
    ConAnuncios("Nerea", 12),
]

for s in suscripciones:
    print(s.ficha())

ingresos = sum(s.precio_mensual() for s in suscripciones)
print(f"Ingresos mensuales: {ingresos:.2f} €")
````

**Pregunta final:** si mañana el restaurante añade una clase `PlatoInfantil` con un precio fijo de 6 €, ¿qué partes del programa tendrías que cambiar y cuáles funcionarían sin tocarlas? Justifica tu respuesta.

````python
from functools import reduce

# ==========================================
# Parte A. Jerarquía de clases
# ==========================================
class Plato:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = float(precio)

    def precio_final(self):
        return self.precio

    def __str__(self):
        return f"{self.nombre} - {self.precio_final():.2f} €"

class PlatoVegano(Plato):
    def precio_final(self):
        return round(super().precio_final() * 0.90, 2)

class PlatoGourmet(Plato):
    def precio_final(self):
        return round(super().precio_final() + 3.0, 2)


# ==========================================
# Parte B. Datos del día
# ==========================================
pedidos = [
    {"cliente": "Ana",   "plato": PlatoVegano("Ensalada quinoa", 11.0),  "unidades": 2},
    {"cliente": "Luis",  "plato": Plato("Hamburguesa", 9.5),             "unidades": 1},
    {"cliente": "Marta", "plato": PlatoGourmet("Risotto trufa", 18.0),   "unidades": 1},
    {"cliente": "Ana",   "plato": Plato("Pizza margarita", 10.0),        "unidades": 3},
    {"cliente": "Pablo", "plato": PlatoVegano("Curry garbanzos", 12.0),  "unidades": 1},
]


# ==========================================
# Parte C. Comprehensions
# ==========================================
# 50. Lista con el importe de cada pedido
importes_pedido = [round(p["plato"].precio_final() * p["unidades"], 2) for p in pedidos]

# 51. Set con los clientes distintos (lo ordenaremos con sorted() al mostrarlo)
clientes_distintos = {p["cliente"] for p in pedidos}

# 52. Diccionario nombre del plato -> precio final
precios_finales = {p["plato"].nombre: round(p["plato"].precio_final(), 2) for p in pedidos}


# ==========================================
# Parte D. Programación funcional
# ==========================================
# 53. Pedidos de platos veganos con filter() e isinstance()
pedidos_veganos = filter(lambda p: isinstance(p["plato"], PlatoVegano), pedidos)
nombres_veganos = [p["plato"].nombre for p in pedidos_veganos]

# 54. Facturación total con map() y reduce()
facturacion_total = reduce(
    lambda acc, imp: round(acc + imp, 2),
    map(lambda p: p["plato"].precio_final() * p["unidades"], pedidos),
    0
)

# 55. Clausura crear_filtro_minimo(minimo)
def crear_filtro_minimo(minimo):
    def filtro(pedido):
        return (pedido["plato"].precio_final() * pedido["unidades"]) >= minimo
    return filtro

es_pedido_grande = crear_filtro_minimo(20)
clientes_20_o_mas = [p["cliente"] for p in filter(es_pedido_grande, pedidos)]


# ==========================================
# Parte E. Recursión
# ==========================================
def contar_unidades(lista_pedidos):
    # Caso base: si no quedan pedidos en la lista, 0 unidades
    if not lista_pedidos:
        return 0
    # Caso recursivo: unidades del primer pedido + unidades del resto de la lista
    return lista_pedidos[0]["unidades"] + contar_unidades(lista_pedidos[1:])


# ==========================================
# Parte F. Informe final
# ==========================================
print("=== CARTA DEL DÍA ===")
for p in pedidos:
    print(p["plato"])

print(f"\nImportes por pedido: {importes_pedido}")
print(f"Clientes distintos: {sorted(clientes_distintos)}")
print(f"Precios finales: {precios_finales}")
print(f"Platos veganos pedidos: {nombres_veganos}")
print(f"Facturación total: {facturacion_total} €")
print(f"Pedidos de 20 € o más: {clientes_20_o_mas}")
print(f"Unidades vendidas: {contar_unidades(pedidos)}")
````
