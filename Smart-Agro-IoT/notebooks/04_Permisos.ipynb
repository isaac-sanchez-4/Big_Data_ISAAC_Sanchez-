```python
%sql
USE CATALOG agro_catalog;

-- caso administrador control total sobres el catalogo
GRANT ALL PRIVILEGES ON CATALOG agro_catalog TO `admin`;
--Empleados acceso unicamente a la capa de negocio
GRANT USE CATALOG ON CATALOG agro_catalog TO `operarios`;
GRANT USE SCHEMA ON SCHEMA agro_catalog.bronce TO `operarios`;

-- permitimos las consultas en la capa gold
GRANT SELECT ON TABLE agro_catalog.bronce.gold_consumo_agua_parcela TO `operarios`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_estres_suelo TO `operarios`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_resumen_clima_diario TO `operarios`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_resumen_alertas TO `operarios`;

-- bloqueo a capas que no sean de negocio para prevenir cambios en los datos por errores 
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.sensor_readings_bronze FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.sensor_readings_silver FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.weather_readings_bronze FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.weather_readings_silver FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.irrigation_events_bronze FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.irrigation_events_silver FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.alerts_bronze FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.alerts_silver FROM `operarios`;
REVOKE ALL PRIVILEGES ON TABLE agro_catalog.bronce.calidad_rechazos_silver FROM `operarios`;


-- Asistente de ia con garantias de solo poder consultar datos 
-- Damos permiso para que acceda a las catalogos y limitando la lectura a las capa de negocio  
GRANT USE CATALOG ON CATALOG agro_catalog TO `ai_assistant`;
GRANT USE SCHEMA ON SCHEMA agro_catalog.bronce TO `ai_assistant`;

GRANT SELECT ON TABLE agro_catalog.bronce.gold_consumo_agua_parcela TO `ai_assistant`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_estres_suelo TO `ai_assistant`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_resumen_clima_diario TO `ai_assistant`;
GRANT SELECT ON TABLE agro_catalog.bronce.gold_resumen_alertas TO `ai_assistant`;
-- garantizamos que la ia no pueda ralizar cualquier modificacion para prevenir cambios por alucinaciones de la ia
REVOKE INSERT, UPDATE, DELETE ON CATALOG agro_catalog FROM `ai_assistant`;
```
